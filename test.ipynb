```python
def get_prime(n):

    for num in range(2, n + 1):
        is_prime = True
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            yield num
primes=get_prime(10)
for i in primes:
    if i>10:
        break
    print(i)
```

```python
from statistics import mean, median, mode, stdev
def moving_median(data):
    for i in range(len(data)):
        yield median(data[:i+1])
for m in moving_median([1, 3, 5, 7, 9]):
    print(m)

```
